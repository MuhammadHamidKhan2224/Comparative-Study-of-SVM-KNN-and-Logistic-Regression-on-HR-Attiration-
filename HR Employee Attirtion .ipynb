Necessary preprocessing steps—including handling missing values, label encoding, one-hot encoding, and scaling of numerical features—were completed to ensure model reliability. Finally, the dataset was split into training and testing sets using an 80/20 ratio, setting the stage for robust predictive modeling.

### For predictive modeling, the data analyst implemented logistic regression, support vector machine (SVM), and k-nearest neighbours (KNN) classifiers. Initially, logistic regression achieved an accuracy of around 73%, while SVM improved slightly to 78% accuracy but suffered from lower recall. Conversely, the KNN model, after rigorous hyperparameter tuning with both RandomizedSearchCV and GridSearchCV, performed best with an accuracy of 88% and significantly higher precision and recall. In addition, cross-validation provided consistent results and helped refine the optimal parameters for each model, ultimately demonstrating that the tuned KNN model was the most effective at predicting employee attrition.

### Our findings indicate that factors such as frequent business travel, lower work-life balance, minimal promotion opportunities, and reduced compensation contribute to higher attrition rates. Similarly, employees in departments like Human Resources are at greater risk of leaving compared to those in other divisions. Based on these insights, it is recommended that organizations focus on improving employee satisfaction through targeted retention programs, career development initiatives, and mentoring opportunities. In addition, proactive monitoring of workforce trends can help identify at-risk employees, thereby enabling timely interventions that support long-term organizational stability.
